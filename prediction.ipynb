```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import load_model
```

```python
word_index = imdb.get_word_index()
reversed_word_index = {value:key for key,value in word_index.items()}
```

```python
model = tf.keras.models.load_model('simple_rnn_imdb.h5')

model.summary()
```

```python
#Step 2 helper functions:
#function to decode reviews

def decode_review(encoded_review):
    return ' '.join([reversed_word_index.get(i-3,'?') for i in encoded_review])

#Funtion to preprocess the input

def preprocess_text(text):
    words = text.lower().split()
    encoded_reviews = [word_index.get(word,2)+3 for word in words]
    padded_sequence = sequence.pad_sequences([encoded_reviews],maxlen=500)
    return padded_sequence
```

```python
#Prediction function

def predict_sentiment(review):
    predict_input = preprocess_text(review)

    prediction = model.predict(predict_input)

    sentiment = "Positive" if prediction[0][0] > 0.5 else "Negative"
    
    return sentiment,prediction[0][0]
```

```python
eg_review = "This movie was fantastic, I really loved the movie"

sentiment,score = predict_sentiment(eg_review)

print(f' The Review is: {eg_review}')
print(f' The Sentiment predicted is :{sentiment}')
print(f'Prediction Score : {score}')
```
